```python
import sys
sys.path.append('../..')

import joblib
import numpy as np
import pandas as pd
from ml.src.train_anomaly import generate_anomaly_training_data
```

## 1. Load Isolation Forest Model


```python
model = joblib.load('../../ml/artifacts/anomaly_model.joblib')
print("Isolation Forest successfully loaded.")
print(f"Number of Estimators: {model.n_estimators}")
print(f"Contamination Rate:   {model.contamination}")
```

## 2. Test Scoring on Normal vs Anomalous Invoices


```python
# Features: [total_amount_log, tax_ratio, subtotal_to_total_ratio, math_discrepancy]
# 1. Normal invoice: Total = 11,800, Subtotal = 10,000, Tax = 1,800 (18%)
normal_sample = np.array([[np.log1p(11800.0), 1800.0 / 10000.0, 10000.0 / 11800.0, 0.0]])
score_norm = model.decision_function(normal_sample)[0]
pred_norm = model.predict(normal_sample)[0]

# 2. Anomalous tax rate: Tax = 70%
tax_anomaly = np.array([[np.log1p(6800.0), 2800.0 / 4000.0, 4000.0 / 6800.0, 0.0]])
score_anom = model.decision_function(tax_anomaly)[0]
pred_anom = model.predict(tax_anomaly)[0]

# 3. Mathematical discrepancy: Subtotal (8500) + Tax (1530) != Total (12000)
math_anomaly = np.array([[np.log1p(12000.0), 1530.0 / 8500.0, 8500.0 / 12000.0, 1970.0]])
score_math = model.decision_function(math_anomaly)[0]
pred_math = model.predict(math_anomaly)[0]

print(f"Normal Invoice:        Score = {score_norm:+.4f} | Prediction = {pred_norm} (1 = Normal)")
print(f"Tax Rate Anomaly:      Score = {score_anom:+.4f} | Prediction = {pred_anom} (-1 = Anomaly)")
print(f"Math Discrepancy:      Score = {score_math:+.4f} | Prediction = {pred_math} (-1 = Anomaly)")
```
