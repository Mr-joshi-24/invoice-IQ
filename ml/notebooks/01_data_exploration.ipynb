# DocuMind AI — 01. Data Exploration & Layout Analysis
This notebook explores invoice document structures, Indian GST identification syntax, and token-level ground truth distributions.


```python
import sys
sys.path.append('../..')

from ml.datasets.synthetic_generator import generate_dataset, generate_synthetic_invoice_document
import pandas as pd
from collections import Counter
```

## 1. Inspect Single Synthetic Invoice


```python
doc = generate_synthetic_invoice_document('Nexus Digital Systems Inc', doc_index=1)
print(f"Document ID: {doc['document_id']}")
print(f"Vendor: {doc['vendor_name']}")
print(f"Total Tokens: {len(doc['tokens'])}")
print("Ground Truth Fields:")
for k, v in doc['ground_truth'].items():
    print(f"  - {k}: {v}")
```

## 2. Token Class Distribution Analysis


```python
dataset = generate_dataset(total_documents=100)
all_labels = [t['label'] for doc in dataset['train'] for t in doc['tokens']]
label_counts = Counter(all_labels)

df_labels = pd.DataFrame(label_counts.items(), columns=['Token Class', 'Count']).sort_values(by='Count', ascending=False)
print("Token Class Frequencies across Training Split:")
print(df_labels.to_string(index=False))
```

## 3. Indian GSTIN Syntax Verification


```python
import re
GSTIN_REGEX = re.compile(r"^\d{2}[A-Z]{5}\d{4}[A-Z]{1}[1-9A-Z]{1}Z[0-9A-Z]{1}$")

sample_gstins = [doc['ground_truth']['gstin'] for doc in dataset['train'][:10]]
for g in sample_gstins:
    valid = bool(GSTIN_REGEX.match(g))
    state_code = g[:2]
    pan = g[2:12]
    print(f"GSTIN: {g} | State: {state_code} | PAN: {pan} | Valid: {valid}")
```
