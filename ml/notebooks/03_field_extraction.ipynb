## 1. Inspect Trained Classifier & Benchmark Artifacts


```python
with open('../../ml/artifacts/metrics_summary.json', 'r') as f:
    metrics = json.load(f)

print(f"Exact-Match Accuracy: {metrics['exact_match_accuracy'] * 100:.1f}%")
print(f"Weighted F1-Score:    {metrics['weighted_f1']:.4f}")
print(f"Macro F1-Score:       {metrics['macro_f1']:.4f}")
print(f"Overall Accuracy:     {metrics['overall_accuracy'] * 100:.1f}%")
```

## 2. Per-Class Precision & Recall Breakdown


```python
report = metrics['classification_report']
classes = ['DATE', 'GSTIN', 'INVOICE_NUM', 'SUBTOTAL', 'TAX', 'TOTAL', 'VENDOR', 'O']
rows = []
for c in classes:
    if c in report:
        rows.append({
            'Class': c,
            'Precision': round(report[c]['precision'], 3),
            'Recall': round(report[c]['recall'], 3),
            'F1-Score': round(report[c]['f1-score'], 3),
            'Support': report[c]['support']
        })
print(pd.DataFrame(rows).to_string(index=False))
```

## 3. Confusion Matrix


```python
labels = metrics['labels']
cm = pd.DataFrame(metrics['confusion_matrix'], index=labels, columns=labels)
print("Confusion Matrix:")
print(cm)
```
